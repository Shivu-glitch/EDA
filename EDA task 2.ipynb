```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
```

```python
df
```

```python
sns.histplot(df["Sales"],bins=8)
plt.title("Distribution of sales")
plt.show()
```

```python
sns.histplot(df["Profit"],bins=8)
plt.title("Distribution of Profit")
plt.show()
```

```python
sns.boxplot(x=df["Profit"])
plt.title("Boxplot of Profit")
plt.show()
```

```python
sns.scatterplot(x="Discount_Percent",y="Profit_Margin_Percent",data=df)
plt.title("Discount / Profit Margin")
plt.show()
```

```python
sns.barplot(x="Discount_Percent",y="Profit_Margin_Percent",data=df)
plt.title("Discount / Profit Margin")
plt.show()
```

```python
numaric_column = [
    "Transaction_ID",
    "Unit_Price",
    "Quantity",
    "Discount_Percent",
    "Sales",
    "Profit",
    "Profit_Margin_Percent",
    "Cost",
]
correlation = df[numaric_column].corr()
```

```python
sns.heatmap(correlation,
            annot=True,
            cmap="Reds",
            cbar=False,
            square=True,
            linewidth=2)
plt.title("all numerical")
plt.show()
```
