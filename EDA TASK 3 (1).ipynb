```python
#identifying anomlies means errors
Volume_mean=df["Volume"].mean()
Volume_mean
```

```python
#identifying anomlies means errors
Volume_std=df["Volume"].std()
Volume_std
```

```python
Volume_threshold=Volume_mean+(2*Volume_std)
```

```python
anomalies=df[df["Volume"]>Volume_threshold]
anomalies
```

```python
#stock retuen distrubution
mean_return=df["Daily_Return"].mean()
print("mean_return",mean_return)
variance_return=df["Daily_return"].var()
print("variance_return",variance_return)
standerd_return=df["Daily_Return"].std()
print("standard_return",standerd_return)
```
